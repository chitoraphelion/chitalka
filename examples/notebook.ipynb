# Градиентный спуск в ноутбуке

Шаг $x_{k+1} = x_k - \eta\nabla f(x_k)$ при $\eta \le 1/L$.

$$
f(x_k) - f^\star \le \frac{L\|x_0 - x^\star\|^2}{2k}
$$

```python
import numpy as np
x = np.array([3.0, -1.0])
for k in range(3):
    x = x - 0.5 * 2 * x
    print(k, x)
```
