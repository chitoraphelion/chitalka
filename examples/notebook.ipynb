$$
f(x_k) - f^\star \le \frac{L\|x_0 - x^\star\|^2}{2k}
$$

```python
import numpy as np
x = np.array([3.0, -1.0])
for k in range(3):
    x = x - 0.5 * 2 * x
    print(k, x)
```
